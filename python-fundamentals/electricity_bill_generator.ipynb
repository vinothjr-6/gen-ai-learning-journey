```python
#TNEB BILL GENERATOR
name = input("Enter your name: ")
tariff = input("Enter your tariff (domestic/commercial): ")
units = int(input("Enter the number of units consumed: "))

if (tariff == "domestic"):
    if units <= 500:
        if units <= 200:
            total_bill = 0
        elif units <= 300:
            total_bill = ((units - 200) * 3)
        elif units <= 400:
            total_bill = ((100 * 3) + ((units - 300) * 4))
        else:
            total_bill = ((100 * 3) + (100 * 4) + ((units - 400) * 5))
    else:
        total_bill = ((100 * 2.5) + (100 * 3) + (100 * 4) + (100 * 5) + ((units - 500) * 6))
    print("Name:", name)
    print("Tariff:", tariff)
    print("Units Consumed:", units)
    print("Total Bill: Rs.", total_bill)

elif (tariff == "commercial"):
    if units <= 500:
        if units <= 50:
            total_bill = 0
        elif units <= 100:
            total_bill = units * 3
        elif units <= 200:
            total_bill = ((100 * 3) + ((units - 100) * 5))
        elif units <= 300:
            total_bill = ((100 * 3) + (100 * 5) + ((units - 200) * 6))
        elif units <= 400:
            total_bill = ((100 * 3) + (100 * 5) + (100 * 6) + ((units - 300) * 7))
        elif units <= 500:
            total_bill = ((100 * 3) + (100 * 5) + (100 * 6) + (100 * 7) + ((units - 400) * 8))
    else:
        total_bill = ((100 * 3) + (100 * 4.5) + (100 * 5) + (100 * 6) + (100 * 7) + ((units - 500) * 8))
    print("Name:", name)
    print("Tariff:", tariff)
    print("Units Consumed:", units)
    print("Total Bill: Rs.", total_bill)

else:
    print("provide valid tariff type")
    
```
