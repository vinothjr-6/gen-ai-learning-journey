```python
product = 1000
quantity = 2
subtotal = product * quantity
print("subtotal : ",subtotal)
coupon = True
if subtotal >= 1000 and coupon :
    final_total = subtotal - 100
    print("discount eligible")
    print("total : ",final_total)
else:
    final_total = subtotal
    print("Total : ",final_total)
budget = 5000

print("within Budget",budget > final_total)
```
