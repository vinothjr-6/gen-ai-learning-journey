```python
print("WELCOME")
print()

# customer data
c_name = input("Enter your name: ")
c_name = c_name.upper()
c_age = int(input("Enter your age: "))
print()

# menu list
menu_11 = ["Briyani", "fried rice", "noodles"]
length_of_menu = len(menu_11)

#price list
menu = {
    "briyani": 150,
    "fried rice": 120,
    "noodles": 100
}
print("Menu")
print(menu,"\n""Total items in the menu:",length_of_menu)
print()

#food details
f_name = input("Enter the food name: ")
f_name = f_name.lower()
f_quantity = int(input("Enter the quantity: "))
delivery = input("Do you want home delivery? (yes/no): ")
print()

#timeslots tuple
time_slots = ("12:00 PM", "1:00 PM", "2:00 PM")
print("Available time slots:",time_slots)
delivery_time = input("Enter your preferred delivery time slot: ")
print()

#payment methods set
payment_methods = {"cash", "card", "UPI"}
print("Available payment methods:",payment_methods)
payment = input("Enter your preferred payment method: ")
print()

total_cost = menu[f_name] * f_quantity

#oder summary
order_summary = {
    "Customer_Name": c_name,
    "food_item": f_name,
    "quantity": f_quantity,
    "total_cost": total_cost
}
print()

print(order_summary)
print()

print("ORDER SUMMARY")
print()
print("Customer Name:", order_summary["Customer_Name"])
print("Food Item:", order_summary["food_item"])
print("Total Cost:", order_summary["total_cost"])
print("Delivery time:", delivery_time)
print("Payment method:", payment)
```
